```python
#錯誤寫法, 後續延伸要重寫
i = 0
while i < 10:
    print("*" * (i+1))
    i=i+1

layer = 10
i = 0
while i < layer:
    print("p" * (layer-1-i)+ "*" * (i+1))
    i=i+1

#有結構的寫法
left, right = " ", "*"
layer = 10
i = 0
while i < layer:
    lcount = layer-i-1
    rcount = i + 1
    print(left * lcount+ right * rcount)
    i=i+1
```

```python
left, mid, right = " ", "*", " "
layer = 10
i = 0
while i < layer:
    lcount = i
    mcount = (2*layer-1)-2*i
    rcount = i
    print(left * lcount +
          mid * mcount +
          right * rcount)
    i=i+1
```

```python
# 除法計算 2/7 -> 0.2857142857
a, b = 2, 7
ans= "0."
i = 0
while i<10:
    a10 = a*10
    ans = ans + str(a10 // b)
    print (ans)
    a = a10 % b
    i=i+1
```

```python
#目前累積(0) + 這次要加(1) = 記憶總數, 當i沒到10, 繼續迴圈
total = 0
i = 0
while i <10:
    total = total + (i + 1)
    print(total)
    i=i+1
print("!!!!", total)
```

```python
#座標
x, y = 0, 0
commands = "ULLRDL" #x=-2, y=0
i = 0

while i < len(commands):
    c = commands[i]
    if c == "U":
        y = y + 1
    elif c == "D":
        y = y - 1
    elif c == "L":
        x = x - 1
    elif c == "R":
        x = x + 1

    print(x, y)
    i = i + 1
```

```python
from itertools import count

import random
count = 0
total = 1000000
i = 0
while i < total:
    # 對於這次隨機(x, y)
    x = random.uniform(-1,1)
    y = random.uniform(-1,1)
    # 如果這座標在圓裡面, 累積次數+1
    if x**2 + y**2 <= 1:
        count = count + 1
    i = i + 1
# 圓形裡面 / 全部 * 參照物面積
print(count / total * 4)
```

```python
#找最小值
l = [80, 20, 30, 40]

# 一路走過去, 看到更小的值, 就把這個值以及它的位置記錄起來
minv, mini = float("inf"), -1
# 設定最不可能的最大值(ex:99999 或 float("inf") 無限大和不可能的位置

for i in range(len(l)):
    n = l[i]
    if n < minv:
        minv, mini = n, i
    print(minv, mini)
```
